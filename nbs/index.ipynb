# boxrecipe

> The doyu-box composition: SERVICES registry + box recipe

Composes every doyu-box service (Caddy sites, systemd units, install cmds) plus WireGuard and the maintenance baseline into one cloud-init YAML. See `10_box.ipynb` for the recipe and the live rebuild flow.

Library primitives live in [hetznerinit](https://github.com/doyu/hetznerinit).
Design: `docs/superpowers/specs/2026-07-15-boxrecipe-discopipe-design.md`
